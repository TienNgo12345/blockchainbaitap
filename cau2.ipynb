```python
!openssl genrsa -out private_key.pem 2048
!openssl rsa -pubout -in private_key.pem -out public_key.pem
```

```python
!openssl pkeyutl -encrypt -pubin -inkey public_key.pem -in plaintext.txt -out encrypted_rsa.bin
```

```python
!openssl pkeyutl -decrypt -inkey private_key.pem -in encrypted_rsa.bin -out decrypted_rsa.txt
```
